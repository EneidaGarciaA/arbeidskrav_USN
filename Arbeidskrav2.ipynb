Lag et program med en funksjon som tar a og b som inn-argumenter og som så regner ut arealet og «ytre» omkrets til en figur satt sammen av en rettvinklet trekant og en halvsirkel, se figuren under.  Med «ytre» omkrets menes samlet lengde av de sorte strekene. 
Funksjonen skal returnere arealet og «ytre» omkrets, som så skrives til skjerm med passende tekst.  

```python
#Oppgave 5: løsning
import numpy as np
a= float(input("Skriv lengden til katet a:",   ))
b= float(input("Skriv lengden til katet b:",  ))
arealet_trekant = (a * b) /2
arealet_halv_sirkel = (np.pi * a * a) / 2
areal_hele_figuren = arealet_halv_sirkel + arealet_trekant
c= np.sqrt(a**2 + b**2)
omkrets_hele_figuren = (np.pi* a/2) + b + c
print("arealet av trekanten er: ",arealet_trekant)
print("arealet av halvsirkelen er:",round(arealet_halv_sirkel, 2) )
print("omkrets av hele figuren er: ", round(omkrets_hele_figuren, 2))
```

Oppgave 6
Skriv en kode som plotter funksjonen 𝑓(𝑥) = −𝑥**2 − 5, for x på intervallet [-10,10].
Hint: np.linspace(-10, 10, 200) gir en array med 200 punkter jevnt fordelt på intervallet   [-10,10]. 


```python
#oppgave 6: løsningen
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-10, 10, 200)
y = -x**2 - 5
plt.plot(x, y)
plt.plot(color = "blue", linestyle = "-", linewidth = 3)
plt.xlabel("x-aksen")
plt.ylabel("y = f(x)")
plt.title("Grafen til funksjonen (-x**2) - 5")
plt.plot(0, -5, "D", markersize = 8)
plt.grid()
```
